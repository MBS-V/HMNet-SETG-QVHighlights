# Evaluation
This notebook runs the repository's internal single-ground-truth metric implementation. For benchmark-comparable QVHighlights results, run the official evaluator and preserve its protocol.

```python
from pathlib import Path
import os,sys,subprocess
try:
    from google.colab import drive
    drive.mount('/content/drive')
except ImportError:
    pass
candidates=[Path.cwd(),Path('/content/HMNet_SETG'),Path('/content/drive/MyDrive/nlvms/HMNet_SETG')]
ROOT=next((p for p in candidates if (p/'README.md').exists()),None)
if ROOT is None: raise FileNotFoundError('Project folder not found; extract the ZIP first.')
os.chdir(ROOT); sys.path.insert(0,str(ROOT))
MANIFEST=os.environ.get('HMNET_VAL_MANIFEST','/content/drive/MyDrive/nlvms/features/qvhighlights/val.jsonl')
FEATURE_ROOT=os.environ.get('HMNET_FEATURE_ROOT','/content/drive/MyDrive/nlvms/features/qvhighlights')
CHECKPOINT=os.environ.get('HMNET_CHECKPOINT','/content/drive/MyDrive/nlvms/checkpoints/qvh_hmnet_baseline/best.pt')
assert Path(MANIFEST).exists() and Path(CHECKPOINT).exists()
```

```python
cmd=[sys.executable,'scripts/evaluate.py','--config','configs/qvhighlights.yaml','--manifest',MANIFEST,'--feature-root',FEATURE_ROOT,'--checkpoint',CHECKPOINT,'--output-dir','outputs/qvh_validation']
subprocess.run(cmd,check=True)
print('See outputs/qvh_validation/metrics.json and predictions.csv')
```
